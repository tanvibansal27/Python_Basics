```python
#Count how many times a particular number appears in a list.
def count(nums,target):
    count = 0
    for num in nums:
        if(num == target):
            count+=1
    return count
count(nums,50)
```

```python
#Reverse a list without using .reverse() or slicing.
def reverse(nums):
    left = 0
    right = len(nums) - 1
    while left < right:
        nums[left] , nums[right] = nums[right] , nums[left]
        left +=1 
        right -= 1
    return nums
reverse(nums)
```

```python
#Remove duplicate values from:
numbers = [10, 20, 10, 30, 20, 40, 50, 30]
def rmv_duplicate(nums):
    result = []
    for num in nums :
        if num not in result:
            result.append(num)
    return result
rmv_duplicate(numbers)
```

```python
#Find the second-largest number in a list.
def s_largest(nums):
    largest = float("-inf")
    s_largest = float("-inf")
    for num in nums:
        if(num > largest):
            s_largest = largest
            largest = num
        elif(num > s_largest and num != largest):
            s_largest = num
    return s_largest
s_largest(nums)
```

```python
# Count:
# • Letters
# • Numbers
# • Spaces
# in a given string.
text = "Hello Python 123"
letters = 0
nums = 0
spaces = 0

for c in text :
    if c.isalpha() :
        letters += 1
    elif c.isdigit():
        nums += 1
    else : 
        spaces += 1
print ("Letters " , letters)
print("Numbers " , nums)
print("spaces " , spaces)
```
