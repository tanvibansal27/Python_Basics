```python
def add(a,b):
    return a+b
def sub(a,b):
    if(a>b):
        return a-b
    else:
        return b-a
def mul(a,b):
    return a*b
def div(a,b):
    if b==0 :
        return "Divide by zero is not possible"
    else:
        return a/b

print(add(1,2))
print(sub(2,4))
print(mul(3,6))
print(div(2,0))
```

```python
def check_eve_odd(num):
    if num % 2 == 0:
        return "Even"
    else:
        return "Odd"
check_eve_odd(5)
```

```python
def largestOf3_nos(a,b,c):
    if(a>b and a>c) :
        return a
    elif(b>a and b>c):
        return b
    else:
        return c

largestOf3_nos(2,8,4)
```

```python
def square_cube(num):
    print("Square -> " , num**2)
    print("Cube -> " , num ** 3)
square_cube(2)
```

```python
def sumOfList(nums):
    sum = 0
    for num in nums:
        sum += num
    return sum
sumOfList([2,3,4,5])
```

```python
def max_min(nums):
    return max(nums) , min(nums)
max_min([2,3,4,5,98,1,0])
```

```python
def numOfWords(sentence):
    L = sentence.split()
    count = 0
    for word in L:
        count+=1
    return count
numOfWords("I Love me and my family")
```

```python
def check_prime(num):
    for i in range (2 , num):
        if(num % i == 0) :
            return False
    return True
check_prime(7)
```

```python
def Even_List(nums):
    even_list = []
    for num in nums:
        if num % 2 == 0:
            even_list.append(num)
    return even_list
Even_List([2,3,4,5,6,7,8,9,10])
```

```python
def average(*args):
    return sum(args)/len(args)
print(average(2,3,4,5,6))
```

```python
def employee_info(**kwargs):
    for key , value in kwargs.items():
        print(key , ": " , value)
employee_info(name = "Tanvi",
              age = "20"
              )
```

```python
def salary_details(info):
    return sum(info) , sum(info)/len(info) , max(info) , min(info)
salary_details([25000,50000,100000,1000,1000000])
```
